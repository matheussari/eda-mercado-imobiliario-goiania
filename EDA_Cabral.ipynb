### Limpeza e Conversão da Coluna 'PRICE'

A coluna 'PRICE' contém valores monetários em formato de texto (`R$ 430.000`, `Sob consulta`).

**Passos:**
1.  Substituir 'Sob consulta' por `NaN`.
2.  Remover 'R$', '.' e substituir ',' por '.' para formatar como número.
3.  Converter a coluna para o tipo `float`.

```python
# Limpeza e conversão da coluna 'PRICE'
gyn['PRICE'] = gyn['PRICE'].replace('Sob consulta', np.nan)  # Substituir 'Sob consulta' por NaN
gyn['PRICE'] = gyn['PRICE'].str.replace('R$', '', regex=False) # Remover 'R$'
gyn['PRICE'] = gyn['PRICE'].str.replace('.', '', regex=False)  # Remover separador de milhares '.'
gyn['PRICE'] = gyn['PRICE'].str.replace(',', '.', regex=False)  # Substituir ',' por '.' para decimais
gyn['PRICE'] = pd.to_numeric(gyn['PRICE'], errors='coerce') # Converter para numérico, forçando erros para NaN

print("Coluna 'PRICE' convertida:")
display(gyn[['PRICE']].head())
```

### Limpeza e Conversão das Colunas 'CONDOMÍNIO' e 'IPTU'

As colunas 'CONDOMÍNIO' e 'IPTU' também contêm valores monetários em formato de texto.

**Passos:**
1.  Remover 'R$', '.' e substituir ',' por '.' para formatar como número.
2.  Converter as colunas para o tipo `float`.

```python
# Limpeza e conversão da coluna 'CONDOMÍNIO'
gyn['CONDOMÍNIO'] = gyn['CONDOMÍNIO'].str.replace('R$', '', regex=False)
gyn['CONDOMÍNIO'] = gyn['CONDOMÍNIO'].str.replace('.', '', regex=False)
gyn['CONDOMÍNIO'] = gyn['CONDOMÍNIO'].str.replace(',', '.', regex=False)
gyn['CONDOMÍNIO'] = pd.to_numeric(gyn['CONDOMÍNIO'], errors='coerce')

# Limpeza e conversão da coluna 'IPTU'
gyn['IPTU'] = gyn['IPTU'].str.replace('R$', '', regex=False)
gyn['IPTU'] = gyn['IPTU'].str.replace('.', '', regex=False)
gyn['IPTU'] = gyn['IPTU'].str.replace(',', '.', regex=False)
gyn['IPTU'] = pd.to_numeric(gyn['IPTU'], errors='coerce')

print("Colunas 'CONDOMÍNIO' e 'IPTU' convertidas:")
display(gyn[['CONDOMÍNIO', 'IPTU']].head())
```

### Limpeza e Conversão da Coluna 'AREAS'

A coluna 'AREAS' contém strings como `222 - 485 m²` ou `66 m²`.

**Passos:**
1.  Remover ' m²' e espaços em branco.
2.  Para ranges (ex: '222 - 485'), extrair o primeiro número (inferior).
3.  Converter a coluna para o tipo `float`.

```python
# Limpeza e conversão da coluna 'AREAS'
# Remover ' m²' e extrair o primeiro número de um possível range
gyn['AREAS'] = gyn['AREAS'].str.replace(' m²', '', regex=False).str.strip()
gyn['AREAS'] = gyn['AREAS'].apply(lambda x: str(x).split(' ')[0] if pd.notna(x) else np.nan)
gyn['AREAS'] = pd.to_numeric(gyn['AREAS'], errors='coerce')

print("Coluna 'AREAS' convertida:")
display(gyn[['AREAS']].head())
```

### Limpeza e Conversão das Colunas 'BEDROOMS', 'PARKING-SPACES', 'BATHROOMS'

Essas colunas contêm strings que podem ser números únicos ou ranges (ex: `3 - 4`).

**Passos:**
1.  Para ranges, extrair o primeiro número (inferior).
2.  Converter as colunas para o tipo `float` (para lidar com `NaN`s) e depois para `Int64` (tipo inteiro que aceita `NaN`).

```python
# Funções auxiliares para extrair o primeiro número de um range e converter para numérico
def extract_first_number(value):
    if pd.isna(value):
        return np.nan
    s_value = str(value).strip()
    if '-' in s_value:
        return pd.to_numeric(s_value.split('-')[0].strip(), errors='coerce')
    return pd.to_numeric(s_value, errors='coerce')

columns_to_convert_int = ['BEDROOMS', 'PARKING-SPACES', 'BATHROOMS']

for col in columns_to_convert_int:
    gyn[col] = gyn[col].apply(extract_first_number)
    # Converter para float primeiro para manter NaN, depois para Int64
    gyn[col] = gyn[col].astype('float').astype(pd.Int64Dtype())

print("Colunas de quantidade convertidas:")
display(gyn[columns_to_convert_int].head())
```

### Análise de Média e Mediana para Imputação de Valores Ausentes

Antes de preencher os valores ausentes, vamos calcular a média e a mediana para as colunas numéricas que ainda contêm `NaN`s. A comparação entre esses dois valores nos ajuda a entender a simetria da distribuição dos dados:
- Se a média e a mediana são próximas, a distribuição é mais simétrica e a média pode ser uma boa opção.
- Se há uma diferença significativa, indicando uma distribuição assimétrica (enviesada), a mediana é geralmente uma escolha mais robusta para evitar distorções causadas por *outliers*.

```python
cols_to_impute_numerical = [
    'PRICE', 'AREAS', 'BEDROOMS', 'PARKING-SPACES', 'BATHROOMS', 'CONDOMÍNIO', 'IPTU'
]

# Criar um DataFrame para armazenar as estatísticas
stats_df = pd.DataFrame(index=cols_to_impute_numerical, columns=['Média', 'Mediana'])

print("Média e Mediana das colunas numéricas com valores ausentes:")
for col in cols_to_impute_numerical:
    if gyn[col].isnull().any(): # Apenas se houver valores ausentes
        mean_val = gyn[col].mean()
        median_val = gyn[col].median()
        stats_df.loc[col, 'Média'] = mean_val
        stats_df.loc[col, 'Mediana'] = median_val
    else:
        stats_df.loc[col, 'Média'] = 'N/A (sem NaN)'
        stats_df.loc[col, 'Mediana'] = 'N/A (sem NaN)'

display(stats_df.dropna(how='all')) # Exibe apenas as colunas que realmente precisavam de cálculo
```

### Estratégia de Imputação de Valores Ausentes

Com base na análise de média e mediana e na sua preferência pela mediana em casos de distribuições assimétricas, usaremos a seguinte estratégia:

*   Para as colunas **`PRICE`, `AREAS`, `CONDOMÍNIO`, `IPTU`, `BEDROOMS`, `PARKING-SPACES`, `BATHROOMS`**, onde a mediana é mais robusta contra *outliers* do que a média, preencheremos os valores ausentes com a **mediana**.
*   Para a coluna **`ADDRESS`**, que é um dado textual e a média/mediana não se aplicam, ela será mantida com os valores ausentes por enquanto, pois a imputação simples com a moda pode não ser adequada para seu contexto (endereços).


```python
# Imputando valores ausentes com a mediana para as colunas numéricas
for col in cols_to_impute_numerical:
    if gyn[col].isnull().any():
        median_val = gyn[col].median()
        gyn[col].fillna(median_val, inplace=True)
        print(f"Valores ausentes da coluna '{col}' preenchidos com a mediana: {median_val}")

print("\nVerificando valores ausentes após a imputação:")
display(gyn.isnull().sum())

print("\nInformações atualizadas do DataFrame após a imputação:")
gyn.info()

print("\nPrimeiras 5 linhas do DataFrame após a imputação:")
display(gyn.head())
```
